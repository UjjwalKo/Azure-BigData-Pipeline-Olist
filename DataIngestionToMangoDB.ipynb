```python
try:
  product_category_df = pd.read_csv("data/product_category_name_translation.csv")
except FileNotFoundError:
  print("Error: 'product_category_name_translation.csv' not found.")
  exit() 

host = os.getenv("mango_host")
database = os.getenv("mango_db")
user = os.getenv("mango_user")
password = os.getenv("mango_password")
port = os.getenv("mango_port")

uri = "mongodb://" + user + ":" + password + "@" + host + ":" + port + "/" + database

try:
    client = MongoClient(uri)
    db = client[database]

    collection = db["product_categories"]

    data_to_insert = product_category_df.to_dict(orient="records")
    collection.insert_many(data_to_insert)

    print("Data uploaded to MongoDB successfully!")

except Exception as e:
    print(f"An error occurred: {e}")

finally:
    if client:
        client.close()
     
```
