```python
from pyspark.sql.functions import col, to_date, datediff, current_date, when

def clean_dataframe(df):
    return df.dropDuplicates().drop('all')

orders_df = clean_dataframe(orders_df)
display(orders_df)
```

```python
orders_df = orders_df.withColumn("order_purchase_timestamp", to_date(col("order_purchase_timestamp")))\
    .withColumn("order_approved_at", to_date(col("order_approved_at")))\
    .withColumn("order_delivered_carrier_date", to_date(col("order_delivered_carrier_date")))\
        .withColumn("order_delivered_customer_date", to_date(col("order_delivered_customer_date")))\
            .withColumn("order_estimated_delivery_date", to_date(col("order_estimated_delivery_date")))
```

#### Calculating Delivery and time delays

```python
orders_df = orders_df.withColumn("actual_delivery_time", datediff(col("order_delivered_customer_date"), col("order_purchase_timestamp")))
orders_df = orders_df.withColumn("estimated_delivery_time", datediff(col("order_estimated_delivery_date"), col("order_purchase_timestamp")))
orders_df = orders_df.withColumn("delay time", col("actual_delivery_time") - col("estimated_delivery_time"))
```

```python
orders_customer = orders_df.join(customer_df, orders_df.customer_id == customer_df.customer_id, "left")
order_customer_payment = orders_customer.join(order_payments_df, orders_customer.order_id == order_payments_df.order_id, "left")
order_customer_payment_items = order_customer_payment.join(order_items_df, "order_id", "left")
order_customer_payment_items_products = order_customer_payment_items.join(products_df, order_customer_payment_items.product_id == products_df.product_id, "left")
order_customer_payment_items_products_sellers = order_customer_payment_items_products.join(sellers_df, order_customer_payment_items.seller_id == sellers_df.seller_id, "left")
final_dataframe = order_customer_payment_items_products_sellers.join(mango_spark_data, "product_category_name", "left")
```

```python
def dropping_duplicates_columns(dataframe):
    col = dataframe.columns
    present_col = set()
    to_be_dropped = []
    for c in col:
        if c in present_col:
            to_be_dropped.append(c)
        else:
            present_col.add(c)
    dup_col_free_df = dataframe.drop(*to_be_dropped)
    
    return dup_col_free_df

final_dataframe = dropping_duplicates_columns(final_dataframe)
```

```python
final_dataframe.write.mode("overwrite").parquet("abfss://olistdatacontainer@datastorageolist.dfs.core.windows.net/silver/")
```
