```python
import pandas as pd
import mysql.connector
from mysql.connector import Error
load_dotenv()

db_config = {
    "host": os.getenv("host"),
    "database": os.getenv("database"),
    "user": os.getenv("user"),
    "password": os.getenv("password"),
    "port": os.getenv("port")
}
csv_file = os.getenv("csv_file")    # data/something.csv
table_name = os.getenv("table_name")

def get_sql_type(dtype):
    """Map pandas dtype to MySQL type."""
    if "int" in str(dtype):
        return "INT"
    elif "float" in str(dtype):
        return "FLOAT"
    else:
        return "VARCHAR(255)"

try:
    df = pd.read_csv(csv_file)
    print(f"CSV loaded with {len(df)} rows and {len(df.columns)} columns.")
    connection = mysql.connector.connect(**db_config)
    cursor = connection.cursor()

    cursor.execute(f"DROP TABLE IF EXISTS {table_name};")
    print(f"Table dropped if existed.")

    columns = ", ".join([f"{col} {get_sql_type(dtype)}" for col, dtype in zip(df.columns, df.dtypes)])
    create_query = f"CREATE TABLE {table_name} ({columns});"
    cursor.execute(create_query)
    print(f"Table created successfully.")

    placeholders = ", ".join(["%s"] * len(df.columns))
    insert_query = f"INSERT INTO {table_name} ({', '.join(df.columns)}) VALUES ({placeholders});"

    batch_size = 500
    total_records = len(df)
    for start in range(0, total_records, batch_size):
        batch = df.iloc[start:start+batch_size].values.tolist()
        cursor.executemany(insert_query, batch)
        connection.commit()
        print(f"Inserted {start+1} to {min(start+batch_size, total_records)} records.")

    print(f"All {total_records} records inserted into `{table_name}`.")

except Error as e:
    print("Error while connecting to MySQL:", e)
finally:
    if 'cursor' in locals():
        cursor.close()
    if 'connection' in locals() and connection.is_connected():
        connection.close()
        print("MySQL connection closed.")
```
